 Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def convert_number(number, type_inp, type_out):
    output = 0
    if (type_inp == 'binary' and type_out == 'decimal'):
        number = str(number)
        number = number[::-1]
        for i in range(len(number)):
            output += int(number[i]) * 2**i
        return output
    
    if (type_inp == 'decimal' and type_out == 'binary'):
        converted_number = []
        while (number > 0):
            converted_number.append(str(number%2))
            number = number // 2
        converted_number.reverse()
        return ''.join(converted_number)
        
    
    
    if (type_inp == 'decimal' and type_out == 'exadecimal'):
        exa_dic = {'A': 10, 'B': 11, 'C': 12, 'D':13, 'E':14, 'F':15}
        converted_number = []
        if (number == 0):
            return '0'
        else:
            while (number > 0):
                converted_number.append(str(number%16))
                number = number // 16
            converted_number.reverse()
            for i in range(len(converted_number)):
                for key in exa_dic:
                    if (converted_number[i] == str(exa_dic[key])):
                        converted_number[i] = key
        return ''.join(converted_number)
    
    if (type_inp == 'exadecimal' and type_out == 'decimal'):
        exa_dic = {'A': 10, 'B': 11, 'C': 12, 'D':13, 'E':14, 'F':15}
        number = str(number)
        number = number[::-1]
        output = 0
        for i in range(len(number)):
            if number[i] in exa_dic.keys():
                value = exa_dic[number[i]]
                output += int(value) * 16**i
            else:
                output += int(number[i]) * 16**i
        return output

        

print(convert_number(111,'binary','decimal'))
print(convert_number(7,'decimal','binary'))
print(convert_number(15,'decimal','exadecimal'))
print(convert_number('A','exadecimal','decimal'))
print(convert_number('10000101','binary','decimal'))
print(convert_number('01010100100000000000000','binary','decimal'))
```

2. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
word = 0b01000010101010100100000000000000 #0000 1000 0101 AND 
                                           
def convert_number(word):
    sign = word >> 31 
    exponent = (word >> 23) & 0x0FF
    mantissa = word & 0x007FFFFF
    return (-1)**sign * (1+(mantissa/2**23)) * 2**(exponent - 127)
    

print(convert_number(word))
```
