1\. Create a random list of number and then save it to a text file named "simple_data.txt"

```python
numbers = np.random.rand(10,)
fileName = "simple_data.txt"
with open(fileName, mode = "w") as f:
    for i in numbers:
        f.write(str(i))
        f.write("\n")
```

2\. Create a random matrix of 5x5 and then save it to a text file named "data.txt"

```python
matrix = np.random.rand(25,).reshape(5,5)
fileName = "data.txt"
with open(fileName, mode = "w") as f:
    for i in range(matrix.shape[0]):
        if (i > 0):
            f.write("\n")
        for j in range(matrix.shape[1]):
            f.write(str(matrix[i,j]))
            if j<(matrix.shape[1]-1):
                f.write(" ")
            
```

3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python
fileName = "data.txt"
saveFile = "data.csv"
with open(fileName, mode = 'r') as f:
    with open(saveFile, mode='w') as f2:
        for line in f:
            f2.write(line.replace(" ", ","))
        
            
            
```

5\. Load the file "user_data.json", filter the data by the "CreditCardType" field equals to "American Express". Than save the data a to CSV.

```python
import json
import csv

data = json.load(open('user_data.json'))
#print(data, type(data))
filtered_data = []

for dictionary in data:
    for key,item in dictionary.items():
        if item == "American Express":
            filtered_data.append(dictionary)
print(filtered_data)

fileName = "user_data.csv"
with open(fileName, mode='w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=filtered_data[0].keys())
    writer.writeheader()       
    writer.writerows(filtered_data)
```

7\. Load the remote file [https://www.dropbox.com/s/vkl89yce7xjdq4n/regression_generated.csv?dl=1](https://www.dropbox.com/s/vkl89yce7xjdq4n/regression_generated.csv?dl=1) with Pandas and plot a scatter plot all possible combination of the following fields:
    
  + features_1
  + features_2
  + features_3

```python
url = "https://www.dropbox.com/s/vkl89yce7xjdq4n/regression_generated.csv?dl=1"
data = pd.read_csv(url)
data.head()
fields = ['features_1', 'features_2', 'features_3']

# Loop through all combinations of the selected fields
for i in range(len(fields)):
    for j in range(i + 1, len(fields)):  # Avoid self-combinations
        plt.figure(figsize=(6, 4))
        plt.scatter(data[fields[i]], data[fields[j]], alpha=0.7)
        plt.xlabel(fields[i])
        plt.ylabel(fields[j])
        plt.title(f"Scatter plot: {fields[i]} vs {fields[j]}")
        plt.show()

    
```
