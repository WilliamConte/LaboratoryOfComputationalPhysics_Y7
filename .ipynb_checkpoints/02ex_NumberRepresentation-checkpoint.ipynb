1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def convert_number(number,type_input,type_output):
    
    converted_number = []
    if (type_input == "dec" and type_output == "bin"):
        
        while number > 0:
            converted_number.append(str(number % 2))
            number = number//2
        converted_number.reverse()
        
        return "0b" + ''.join(converted_number)

    
    exa_converter = []
    if (type_input == "dec" and type_output == "exa"):
        hex_map = {10: 'A', 11: 'B', 12: 'C', 13: 'D', 14: 'E', 15: 'F'}
        
        while number > 0:
            remainder = number % 16
            
            if remainder >= 10:
                exa_converter.append(hex_map[remainder])
            else:
                exa_converter.append(str(remainder))
            number = number // 16
        
        exa_converter.reverse()
        return "0x" + ''.join(exa_converter)
    
    if type_input == "bin" and type_output == "dec":
        number = str(number)
        number2 = number[2:]
        sum = 0
        for i in range(len(number2)):
            sum += int(number2[i]) * 2**((len(number2)-1)-i)
        return sum
    
    if type_input == "exa" and type_output == "dec":
        number = str(number)
        str(number)
        number = number[2:]
        hex_map = {'A': 10, 'B': 11, 'C': 12, 'D': 13, 'E': 14, 'F': 15}
        converted_number = []
        for i in number:
            if i in hex_map:
                converted_number.append(hex_map[i])
            else:
                converted_number.append(i)
    
        sum_value = 0
        for j in range(len(number)):
            sum_value += converted_number[j] * (16 ** ((len(converted_number) - 1) - j))
        
        return sum_value
            
    else: 
        print("something went wrong")
        return None
    
print ("10 in binary = ", convert_number(10,"dec","bin"))
print ("10 in exadecimal = ", convert_number(10,"dec","exa"))
print("1010 in decimal = ", convert_number("0b1010","bin","dec"))
print("A in decimal = ", convert_number("0xA","exa","dec"))
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def floating_conversion(number):
    sign = (number >> 31) & 0x8
    exp = (number >> 23) & 0x0ff 
    mantissa = number & 0x007fffff
    return (-1)**sign * 2**(exp - 127) * (1+(mantissa / (1 << 23))) 
    
word = 0b01000010101010100100000000000000
#print(word >> 23)
# exp     10000101
print(floating_conversion(word))


    
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
a, b = 1.0, 1.0
count_a, count_b = 0, 0
underflow_reached = False
overflow_reached = False

while True:
    
    a /= 2
    b *= 2
     
    count_a += 1
    count_b += 1
    
    print("a = ", a, " b = ", b)

    if a == 0.0:
        print("Underflow reached for a.")
        underflow_reached = True
        count_a -= 1

    if b == float('inf'):
        print("Overflow reached for b.")
        overflow_reached = True
        count_b -= 1

    if underflow_reached and overflow_reached:
        break

# Display final counts
print("\nTotal halvings until underflow:", count_a)
print("Total doublings until overflow:", count_b)
```

4\. Write a program to determine the machine precision

**Tips**: define a new variable by adding a smaller and smaller value (proceeding similarly to prob. 2) to an original variable and check the point where the two are the same 

```python
a,b = 0,1
c = 0.5
counter_steps = 0

while True:
    a += c
    c /= 2
    print("a = ", a, "b = ", b)
    if (a != b):
        counter_steps += 1
    else:
        counter_steps += 0
        break
print("iteration where a and b are the same: ", counter_steps)
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute?
