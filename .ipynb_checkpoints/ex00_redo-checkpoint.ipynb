1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
for i in range(1,101):
    if (i%5 == 0 and i%3 == 0):
        print("MickeyMouse")
    elif (i%5 == 0):
        print("Mouse")
    elif (i%3 == 0):
        print("Mickey")
    else:
        print(i)
        
#b
tup = []
for i in range(1,101):
    if (i%5 == 0 and i%3 == 0):
        tup.append("MickeyMouse")
    elif (i%5 == 0):
        tup.append("Mouse")
    elif (i%3 == 0):
        tup.append("Mickey")
    else:
        tup.append(str(i))

for i in range(len(tup)):
    if tup[i] == "Mickey":
        tup[i] = "Donald"
    elif tup[i] == "Mouse":
        tup[i] = "Duck"
tup = tuple(tup)
print(tup)
```

2. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable


```python
def swap (a,b):
    temp = b
    b = a
    a = temp
    return a,b

a,b = 3,5
print(swap(a,b))

def swap2 (a,b):
    return b,a

print(swap2(a,b))
```

3. Computing the distance

Write a function that calculates and returns the euclidean distance between two points u and v, where u and v are both 2-tuples (x,y). For example, if u=(3,0) and v=(0,4), the function should return 5


```python
def distance (u,v):
    u = list(u)
    v = list(v)
    return ((u[0]+v[0])**2 + ((u[1]+v[1])**2))**0.5
u = (3,0)
v = (0,4)
print(distance(u,v))
```

4. Counting letters

Write a program to calculate the number of times each character occurs in a given string s. Ignore differneces in capitalization


```python
string = "stringa di prova"
dic_counter = {}
for i in string:
    if i in dic_counter:
        dic_counter[i] += 1
    else:
        dic_counter[i] = 1
print(dic_counter)
```

5. Isolating the unique

Write a function that determines and count the unique numbers in the list l


```python
l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]

def unique_counter (l):
    dic_counter = {}
    uniques = []
    for i in l:
        if i in dic_counter:
            dic_counter[i] += 1
        else:
            dic_counter[i] = 1
    
    for j in dic_counter:
        if dic_counter[j] == 1:
            uniques.append(j)
    return uniques

print(unique_counter(l))
```

6. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.


```python
def square(a): return a**2
def cube(b): return b**3
def sixth(c): return cube(square(c))

print(square(2), cube(2), sixth(2))
```

8. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem 𝑎2+𝑏2=𝑐2
. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
triples = []
for a in range(1,100):
    for b in range(a,100):
        for c in range(b,100):
            if a**2 + b**2 == c**2:
                triples.append((a,b,c))
print(triples)

triples2 = [(a,b,c) for a in range(1,100) for b in range(a,100) for c in range(b,100) if a**2 +b**2 == c**2]
print(triples2)
```

9. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one


```python
def Norm(mytuple):
    l = list(mytuple)
    partial_sum = 0
    for i in range(len(l)):
        partial_sum += l[i]
    return tuple(l[i]/partial_sum for i in range(len(l)))

prova = (1,2,3,4)
print(Norm(prova))
```
